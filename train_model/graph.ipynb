## accuracy graph

```python
models = ['TabNet', 'ResNet50', 'MobileNet', 'VGG16', 'ViT', 'Swin Transformer']
accuracies = [0.98, 0.92, 0.90, 0.89, 0.91, 0.94]

sns.set_theme(style='whitegrid')
plt.figure(figsize=(10, 6))
barplot = sns.barplot(x=models, y=accuracies, palette='viridis')
for i, acc in enumerate(accuracies):
    barplot.text(i, acc + 0.005, f"{acc:.2f}", ha='center', va='bottom', fontweight='bold')

plt.title('', fontsize=16)
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)  
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
