```python
import pandas as pd
```

```python
def process_csv(input_file, output_file):
    # Đọc tệp CSV
    df = pd.read_csv(input_file)
    
    # Giữ nguyên các dòng có nhãn BENIGN, MSSQL, LDAP, SYN
    labels_to_keep = ["BENIGN", "MSSQL", "LDAP", "Syn"]
    df_filtered = df[df['Label'].isin(labels_to_keep)]
    
    # Xử lý NetBIOS và Portmap (50% mỗi loại)
    netbios = df[df['Label'] == 'NetBIOS'].sample(frac=0.5, random_state=42)
    portmap = df[df['Label'] == 'Portmap'].sample(frac=0.5, random_state=42)
    netbios_portmap_merged = pd.concat([netbios, portmap])
    netbios_portmap_merged['Label'] = 'NetBIOS-Portmap'
    
    # Xử lý UDP và UDP-Lag (50% mỗi loại)
    udp = df[df['Label'] == 'UDP'].sample(frac=0.5, random_state=42)
    udp_lag = df[df['Label'] == 'UDPLag'].sample(frac=0.5, random_state=42)
    udp_merged = pd.concat([udp, udp_lag])
    udp_merged['Label'] = 'UDP-Combined'
    
    # Gộp tất cả lại
    final_df = pd.concat([df_filtered, netbios_portmap_merged, udp_merged])
    
    # Lưu ra tệp CSV mới
    final_df.to_csv(output_file, index=False)
    print(f"Tạo tệp {output_file} thành công!")

# Gọi hàm xử lý
process_csv('data/csv/cicddos_2019.csv', 'data/csv/cicddos_2019_6_labels.csv')
```
