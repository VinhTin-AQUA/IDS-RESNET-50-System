```python
import pandas as pd
import glob
```

```python

# Danh sách các file CSV (cập nhật đường dẫn nếu cần)
file_paths = glob.glob("data/full_data/03-11/*.csv")  # Ví dụ: "data/*.csv"
```

```python
# Đọc và gộp tất cả các file CSV
df_list = [pd.read_csv(file) for file in file_paths]
df = pd.concat(df_list, ignore_index=True)
```

```python
# Kiểm tra số lượng dữ liệu từng nhãn
label_counts = df[" Label"].value_counts()
label_counts
```

```python
# Lấy mẫu cân bằng: Mỗi nhãn có tối đa 5000 dòng
balanced_df = df.groupby(" Label", group_keys=False).apply(lambda x: x.sample(n=min(len(x), 5000)))
```

```python
# Kiểm tra số lượng sau khi cân bằng
print("\nSố lượng mẫu sau khi cân bằng:")
print(balanced_df[" Label"].value_counts())
```
