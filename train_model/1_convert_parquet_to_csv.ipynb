```python
import pandas as pd
import os
```

```python
source = ["data/CICDDOS2019/train/LDAP-training.parquet",
          "data/CICDDOS2019/train/MSSQL-training.parquet",
          "data/CICDDOS2019/train/NetBIOS-training.parquet",
          "data/CICDDOS2019/train/Portmap-training.parquet",
          "data/CICDDOS2019/train/Syn-training.parquet",
          "data/CICDDOS2019/train/UDP-training.parquet",
          "data/CICDDOS2019/train/UDPLag-training.parquet",
          ]
des = ["data/CICDDOS2019/csv/LDAP-training.csv",
          "data/CICDDOS2019/csv/MSSQL-training.csv",
          "data/CICDDOS2019/csv/NetBIOS-training.csv",
          "data/CICDDOS2019/csv/Portmap-training.csv",
          "data/CICDDOS2019/csv/Syn-training.csv",
          "data/CICDDOS2019/csv/UDP-training.csv",
          "data/CICDDOS2019/csv/UDPLag-training.csv",
          ]


for i in range(0,7):
    df = pd.read_parquet(source[i], engine="pyarrow")
    df.to_csv(des[i], index=False)

# # Đọc file Parquet
# 

# # Ghi ra file CSV
# 
```

```python
# Thư mục chứa các file CSV
folder_path = "data/CICDDOS2019/csv"

# Lấy danh sách tất cả các file CSV trong thư mục
csv_files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]

# Tạo một danh sách để lưu dữ liệu từ các file CSV
dfs = []

# Đọc từng file CSV và thêm vào danh sách
for file in csv_files:
    file_path = os.path.join(folder_path, file)
    df = pd.read_csv(file_path)
    dfs.append(df)

# Hợp nhất tất cả DataFrame thành một
merged_df = pd.concat(dfs, ignore_index=True)

# Lưu lại thành một file CSV duy nhất
merged_df.to_csv("data/CICDDOS2019/csv/_merged_output.csv", index=False)

print(f"Đã hợp nhất {len(csv_files)} file CSV thành 'data/_merged_output.csv'")
```
